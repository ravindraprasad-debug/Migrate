```python
from pyspark.sql.functions import *

gold_sales_summary = (
    spark.table("pfl_learning.silver.orders")
    .filter(col("order_status") == "COMPLETED")
    .groupBy("shipping_city")
    .agg(
        count("order_id").alias("total_orders"),
        round(sum("total_amount"), 2).alias("total_sales")
    )
)

gold_sales_summary.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("pfl_learning.gold.sales_summary")

display(gold_sales_summary)
```

```python
from pyspark.sql.functions import *

gold_customer_sales = (
    spark.table("pfl_learning.silver.orders").alias("o")
    .join(
        spark.table("pfl_learning.silver.customers").alias("c"),
        "customer_id"
    )
    .filter(col("order_status") == "COMPLETED")
    .groupBy(
        "customer_id",
        "first_name",
        "last_name",
        "customer_segment"
    )
    .agg(
        count("order_id").alias("total_orders"),
        round(sum("total_amount"), 2).alias("total_sales")
    )
)

gold_customer_sales.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("pfl_learning.gold.customer_sales")
```

```python
from pyspark.sql.functions import *

gold_product_sales = (
    spark.table("pfl_learning.silver.order_items").alias("oi")
    .join(
        spark.table("pfl_learning.silver.products").alias("p"),
        "product_id"
    )
    .groupBy(
        "product_id",
        "product_name",
        "category"
    )
    .agg(
        sum("quantity").alias("quantity_sold"),
        round(sum("line_amount"), 2).alias("total_sales")
    )
)

gold_product_sales.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("pfl_learning.gold.product_sales")
```
